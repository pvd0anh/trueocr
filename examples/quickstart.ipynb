# TrueOCR

Cài `pip install "trueocr[notebook]"` trong môi trường của kernel. Model đã kèm package; load một lần rồi dùng lại.

Ví dụ dùng `sample.png` trong thư mục notebook. Đổi đường dẫn để thử ảnh khác.

```python
from trueocr import load_model

model = load_model()
```

```python
result = model.predict("sample.png")
result.show()
```

Dùng `mode="line"` khi ảnh đã crop sát một dòng. Lấy text qua `result.text`, từng dòng qua `result.lines`; lưu bằng `result.save("output")`.
